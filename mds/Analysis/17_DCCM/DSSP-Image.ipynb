```python
import numpy as np
import matplotlib.pyplot as plt

# Load your data
data = np.loadtxt("dssp_y.dat")
resid = data[:, 0]
extended = data[:, 1]
bridge = data[:, 2]
three = data[:, 3]
alpha = data[:, 4]
pi = data[:, 5]
turn = data[:, 6]
bend = data[:, 7]

# Compute helix and other
helix = [0] * len(pi)
for i in range(len(pi)):
    b = three[i]
    c = alpha[i]
    d = pi[i]
    helix[i] = b + c + d

other = [0] * len(pi)
for i in range(len(pi)):
    e = bridge[i]
    f = turn[i]
    g = bend[i]
    other[i] = e + f + g

# Set font properties to a general Roman font
plt.rcParams.update({
    'font.size': 10, 
    'font.family': 'serif',  
    'font.serif': ['Times New Roman', 'DejaVu Serif', 'Liberation Serif', 'Nimbus Roman No9 L', 'FreeSerif'],  
    'font.weight': 'bold', 
    'axes.labelweight': 'bold'
})

# Create subplots
fig, ax = plt.subplots(3, 1, sharex=True, sharey=True)

# Upper line in gold color with matching label
ax[0].plot(resid, extended, label='Extended', color='gold')
ax[0].set_ylabel('Extended', color='gold')

# Middle line in red color with matching label
ax[1].plot(resid, helix, label='Helix', color='red')
ax[1].set_ylabel('Helix', color='red')

# Lower line in blue color with matching label
ax[2].plot(resid, other, label='Other', color='blue')
ax[2].set_ylabel('Other', color='blue')

fig.text(0.52, 0.01, 'Residues', ha='center')
fig.text(0.001, 0.5, 'Percentage', va='center', rotation='vertical')

plt.tight_layout()
plt.savefig('./ss_per_res_subplots.png', dpi=600)
```
